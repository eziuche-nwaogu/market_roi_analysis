# Simple Linear Regression – Marketing ROI Analysis

## Objective
Analyze marketing data using Simple Linear Regression to determine which advertising channel most influences Sales.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy.stats import shapiro
import warnings

warnings.filterwarnings('ignore')
```

## Load Dataset

```python
marketing_df = pd.read_csv("marketing_data.csv")

marketing_df.head()
```

## Explore Dataset

```python
print(marketing_df.shape)

marketing_df.info()

marketing_df.describe()
```

## Handle Missing Values

```python
print(marketing_df.isnull().sum())

marketing_df = marketing_df.dropna()
```

## Exploratory Data Analysis

```python
sns.pairplot(marketing_df)
plt.show()
```

correlation_matrix = marketing_df.corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.show()

## Select Best Predictor

```python
sales_corr = correlation_matrix['Sales'].drop('Sales')

best_predictor = sales_corr.abs().idxmax()

print(best_predictor)
```

## Build OLS Regression Model

```python
X = marketing_df[[best_predictor]]

y = marketing_df['Sales']

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())
```

## Predictions

```python
marketing_df['Predicted_Sales'] = model.predict(X)

marketing_df[['Sales', 'Predicted_Sales']].head()
```

## Regression Visualization

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=marketing_df[best_predictor],
    y=marketing_df['Sales']
)

sns.lineplot(
    x=marketing_df[best_predictor],
    y=marketing_df['Predicted_Sales'],
    color='red'
)

plt.show()
```

## Diagnostic Tests

```python
residuals = model.resid

fitted_values = model.fittedvalues
```

```python
sns.scatterplot(
    x=fitted_values,
    y=residuals
)

plt.axhline(y=0, color='red', linestyle='--')

plt.show()
```

```python
sns.histplot(residuals, kde=True)

plt.show()
```

```python
sm.qqplot(residuals, line='45')

plt.show()
```

```python
stat, p_value = shapiro(residuals)

print(stat)
print(p_value)
```

## Business Interpretation

TV advertising shows the strongest relationship with Sales.

The regression coefficient indicates that increasing TV advertising spending increases Sales.

The low p-value confirms statistical significance.

Therefore, the company should prioritize TV advertising budget allocation to maximize ROI.
